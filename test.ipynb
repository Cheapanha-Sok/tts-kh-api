```python
import requests

GITLAB_URL = 'http://gdk.test:3000'
ACCESS_TOKEN = "glpat-TOKEN"
def download_file_from_gitlab_model_registry(project_id, model_version_id, file_path, output_file):
    """
    Download a file from GitLab using an access token.

    :param project_id: The ID of the project.
    :param model_version_id: The ID of the model version.
    :param file_path: The path to the file within the project.
    :param output_file: The path to save the downloaded file.
    """
    url = f"{GITLAB_URL}/api/v4/projects/{project_id}/packages/ml_models/{model_version_id}/files/{file_path}"
    headers = {
        "Authorization": f"Bearer {ACCESS_TOKEN}"
    }

    response = requests.get(url, headers=headers)

    if response.status_code == 200:
        with open(output_file, 'wb') as file:
            file.write(response.content)
        print(f"File downloaded successfully to {output_file}")
    else:
        print(f"Failed to download file. Status code: {response.status_code}")

# Example usage:
project_id = 30
model_version_id = 256
file_path = "a.txt"
output_file = "foo1.txt"

download_file_from_gitlab_model_registry(project_id, model_version_id, file_path, output_file)

```
